```python
import operators
```

```python
g = 3
N_list= [10]
w_list = np.linspace((g*0.3), (g*1.5), 200)
m = 0

mu_list = np.linspace(0, (g*1.4), 196)
T_list = np.linspace(0.1, (g*2.5), 196)
```

```python
import csv
from pandas import*
```

```python
def sig(x):
    return 1/(1 + np.exp(-x))
```

```python
for N in N_list:
    for w in w_list:
        print(w)
        chiral_mat = []
        data = []
        for mu in (mu_list):
            chiral_list = []
            H = operators.construct_Hamiltonian(g, w, N, mu, m)
            for T in (T_list):
                rho = operators.get_exact_Gibbs_state(H, T, g)
                chiral = sig(operators.est_chiral(N,rho,(1/(2*w))))
                chiral_list.append(chiral)
                onedata = [(mu/g), (T/g), chiral]
                data.append(onedata)
            chiral_mat.append(np.copy(chiral_list))

        header = ['mu/g', 'T/g', 'chiralcon']
        filename = 'N'+str(N)+'w'+str(w)+'.csv'
        filepath = 'Schwinger Data\\'
        filefull = filepath+filename

        with open(filefull, 'w', newline="") as file:
             csvwriter = csv.writer(file)
             csvwriter.writerow(header)
             csvwriter.writerows(data)

        csvdata = read_csv(filefull)
        npchiral_list = csvdata['chiralcon'].tolist()
        npchiral_array = np.empty((3,196,196))
        k = 0
        
        for i in range(0,196):
            for j in range(0,196):
                npchiral_array[0][i][j] = npchiral_list[k]
                k += 1
        for i in range(0,196):
            for j in range(0,196):
                npchiral_array[1][i][j] = 0
                npchiral_array[2][i][j] = 0

        npfilepath = 'datasets\\Schwinger\\'
        npfilename = filename[:-4] + '.npy'
        npfilefull = npfilepath + npfilename

        if np.isnan(npchiral_array).any():
            print('Nan data detected')
        elif np.isinf(npchiral_array).any():
            print('Inf data detected')
        else:
            np.save(npfilefull, npchiral_array)
```
