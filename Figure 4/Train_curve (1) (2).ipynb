```python
import csv
import matplotlib.cm as cm
```

```python
from pandas import*
```

```python
csvdata = read_csv('demonstration_data.csv')
```

```python
epoch_list = csvdata['epoch'].tolist()
train_loss_list = csvdata['train_loss'].tolist()
test_loss_list = csvdata['test_loss'].tolist()
test_psnr_list = csvdata['test_psnr'].tolist()
```

```python
import matplotlib.pyplot as plt
```

```python
plt.rcParams['font.size'] = 20
plt.rcParams['font.family'] = 'Times New Roman'
plt.rcParams['savefig.dpi'] = 500

fig, ax1 = plt.subplots(figsize = (8, 6))
cmap = cm.get_cmap('coolwarm', 4)
ax2 = ax1.twinx()
ax2.set_ylim(45, 58)
ax1.set_ylim(0, 0.015)
ax1.plot(epoch_list, train_loss_list, label = 'Train Loss', color = cmap(0))
ax1.plot(epoch_list, test_loss_list, label = 'Test Loss', color = cmap(1))
ax2.plot(epoch_list, test_psnr_list, label = 'Test PSNR', color = cmap(3))
ax1.set_xlabel('Epoch')
ax1.set_ylabel('Loss')
ax2.set_ylabel('PSNR')

lines = ax1.get_lines() + ax2.get_lines()
ax1.legend(lines, [line.get_label() for line in lines], loc = 'right')

ax1.set_title('Train Curve')

#plt.show()
plt.savefig("figure4.png", bbox_inches='tight') 
```
