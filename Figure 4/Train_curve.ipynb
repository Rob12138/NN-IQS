```python
import csv
```

```python
from pandas import*
```

```python
csvdata = read_csv('demonstration_data.csv')
```

```python
epoch_list = csvdata['epoch'].tolist()
train_loss_list = csvdata['train_loss'].tolist()
test_loss_list = csvdata['test_loss'].tolist()
test_psnr_list = csvdata['test_psnr'].tolist()
```

```python
import matplotlib.pyplot as plt
```

```python
fig, ax1 = plt.subplots(figsize = (8, 6))

ax2 = ax1.twinx()
ax2.set_ylim(45, 58)
ax1.set_ylim(0, 0.015)
ax1.plot(epoch_list, train_loss_list, label = 'train loss', color = 'b')
ax1.plot(epoch_list, test_loss_list, label = 'test loss', color = 'g')
ax2.plot(epoch_list, test_psnr_list, label = 'test psnr', color = 'r')
ax1.set_xlabel('epoch')
ax1.set_ylabel('loss', rotation = 0, labelpad = 20)
ax2.set_ylabel('psnr', rotation = 0, labelpad = 20)

lines = ax1.get_lines() + ax2.get_lines()
ax1.legend(lines, [line.get_label() for line in lines], loc = 'right')

ax1.set_title('Train Curve')

plt.show()
```
