```python
import matplotlib.cm as cm
def setBoxColors(bp):
    cmap = cm.get_cmap('coolwarm', 4)
    setp(bp['boxes'][0], color='black')
    bp['boxes'][0].set(facecolor = cmap(0))
    setp(bp['caps'][0], color='black')
    setp(bp['caps'][1], color='black')
    setp(bp['whiskers'][0], color='black')
    setp(bp['whiskers'][1], color='black')
    setp(bp['medians'][0], color='black')

    setp(bp['boxes'][1], color='black')
    bp['boxes'][1].set(facecolor = cmap(1))
    setp(bp['caps'][2], color='black')
    setp(bp['caps'][3], color='black')
    setp(bp['whiskers'][2], color='black')
    setp(bp['whiskers'][3], color='black')
    setp(bp['medians'][1], color='black')

    setp(bp['boxes'][2], color='black')
    bp['boxes'][2].set(facecolor = cmap(2))
    setp(bp['caps'][2], color='black')
    setp(bp['caps'][3], color='black')
    setp(bp['whiskers'][2], color='black')
    setp(bp['whiskers'][3], color='black')
    setp(bp['medians'][2], color='black')

    setp(bp['boxes'][3], color='black')
    bp['boxes'][3].set(facecolor = cmap(3))
    setp(bp['caps'][2], color='black')
    setp(bp['caps'][3], color='black')
    setp(bp['whiskers'][2], color='black')
    setp(bp['whiskers'][3], color='black')
    setp(bp['medians'][3], color='black')
```

```python
plt.rcParams['font.size'] = 16
plt.rcParams['font.family'] = 'Times New Roman'
plt.rcParams['savefig.dpi'] = 500

cmap = cm.get_cmap('coolwarm', 4)
bp = plt.boxplot(error_list_2_all, positions = [1, 2, 3, 4], sym='', patch_artist = True, widths = 0.6 )
setBoxColors(bp)
bp = plt.boxplot(error_list_3_all, positions = [6, 7, 8, 9], sym='', patch_artist = True, widths = 0.6 )
setBoxColors(bp)
bp = plt.boxplot(error_list_4_all, positions = [11, 12, 13, 14], sym='', patch_artist = True, widths = 0.6 )
setBoxColors(bp)
bp = plt.boxplot(error_list_6_all, positions = [17, 18, 19, 20], sym='', patch_artist = True, widths = 0.6 )
setBoxColors(bp)
bp = plt.boxplot(error_list_8_all, positions = [22, 23, 24, 25], sym='', patch_artist = True, widths = 0.6 )
setBoxColors(bp)
bp = plt.boxplot(error_list_10_all, positions = [27, 28, 29, 30], sym='', patch_artist = True, widths = 0.6 )
setBoxColors(bp)
plt.title('Up-scaling Test with Seen and Unseen Ratios')
plt.xlabel('Up-scaling Ratio')
plt.ylabel('Relative Error')
plt.ylim(0.00, 0.45)
ticks = ['x2', 'x3', 'x4', 'x6', 'x8', 'x10']
tick_positions = [2.5, 7.5, 12.5, 18.5, 23.5, 28.5]
plt.xticks(tick_positions, ticks)
plt.axvline(x = 15.5, color = 'grey', linestyle = '--')
plt.legend(loc = 'upper left', )
legend_colors = [cmap(0), cmap(1), cmap(2), cmap(3)]
legend_labels = ['NN-IQS', 'Bilinear', 'Cubic', 'Bicubic']
legend_handles = [plt.Rectangle((0,0),1,1, color=color) for color in legend_colors]
plt.legend(legend_handles, legend_labels, loc = 'upper left')
#plt.show()
plt.savefig("figure5a.png", bbox_inches='tight') 
```
